```python
conditions=[
    df['Days of Inventory'] < 3,
    (df['Days of Inventory'] >= 3)&
    (df['Days of Inventory'] < 7),
    (df['Days of Inventory'] >= 7)&
    (df['Days of Inventory'] < 30),
    (df['Days of Inventory'] >= 30)
]

choices=['Critical','Low Stock','Healthy','Overstock']

df['Inventory Status']=np.select(conditions,choices,default='No Demand')
```

```python
product_inventory=df.groupby(['Product ID','Store ID','Category']).agg({
    'Units Sold':'sum',
    'Inventory Level':'mean',
    'Demand':'sum',
    'Units Ordered':'sum',
    'Price':'mean'
}).reset_index()
print(product_inventory.head(10))
```

```python
date_count=df.groupby(['Product ID','Store ID'])['Date'].nunique().reset_index(name='Date Count')
print(date_count)
```

No. of date counts per store 

```python
product_inventory=product_inventory.merge(date_count, on=['Product ID','Store ID'], how='left')
```

```python
product_inventory['Avg Daily Demand']=(product_inventory['Demand'] / product_inventory['Date Count']).round(2)
print(product_inventory.head(10))
```

```python
import numpy as np
product_inventory['Days of Inventory']=np.where(
product_inventory['Avg Daily Demand']>0, (product_inventory['Inventory Level'] / product_inventory['Avg Daily Demand']).round(2), np.nan
)
```

```python
product_inventory["Inventory Status"] = np.select(
    [
        product_inventory["Days of Inventory"] < 3,
        product_inventory["Days of Inventory"].between(3, 7, inclusive="left"),
        product_inventory["Days of Inventory"].between(7, 30),
        product_inventory["Days of Inventory"] > 30
    ],
    [
        "Critical",
        "Low Stock",
        "Healthy",
        "Overstock Candidate"
    ],
    default="No Demand"
)
```

```python
critical_products=product_inventory[product_inventory['Inventory Status']=='Critical']
print(critical_products.head(10))
```

Lead time for quantity replenishment


```python
product_inventory["Lead Time"] = 5

product_inventory["Safety Stock"] = (
    product_inventory["Avg Daily Demand"] * 2
)

product_inventory["Reorder Point"] = (
    product_inventory["Avg Daily Demand"] *
    product_inventory["Lead Time"]
    + product_inventory["Safety Stock"]
)
```

```python
product_inventory['Avg Inventory'] = (product_inventory['Inventory Level']/product_inventory['Date Count']).round(2)
```

Suggested reorder quantity out of the replenushment model

```python
product_inventory["Suggested Reorder Qty"] = np.ceil(
    np.maximum(
        product_inventory["Reorder Point"] -
        product_inventory["Avg Inventory"],
        0
    )
)
```

```python
status_counts = (
    product_inventory["Inventory Status"]
    .value_counts()
    .reset_index()
)

status_counts.columns = [
    "Inventory Status",
    "Product Store Count"
]

plt.figure(figsize=(9, 5))

sns.barplot(
    data=status_counts,
    x="Inventory Status",
    y="Product Store Count"
)

plt.title("Product-Store Inventory Status")
plt.xlabel("Inventory Status")
plt.ylabel("Number of Product-Store Combinations")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```
